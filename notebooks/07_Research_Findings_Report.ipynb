```python
import pandas as pd
from google.colab import files

uploaded = files.upload()

verification = pd.read_csv(
    "Zimbabwe_Lithium_Source_Verification.csv"
)

print("ZIMBABWE LITHIUM — EVIDENCE AUDIT SUMMARY")
print("=" * 45)

print("Total claims:", len(verification))
print("Projects covered:", verification["Project_Name"].nunique())

print("\nClaims by project:")
print(verification["Project_Name"].value_counts())

print("\nVerification results:")
print(verification["Verification_Result"].value_counts())
```

```python
import matplotlib.pyplot as plt

project_counts = verification["Project_Name"].value_counts()

plt.figure(figsize=(10, 5))

project_counts.plot(kind="barh")

plt.title("Zimbabwe Lithium: Claims Documented by Project")
plt.xlabel("Number of Claims")
plt.ylabel("Lithium Project")

plt.tight_layout()
plt.show()
```

```python
# Step 3: Generate a research evidence summary

total = len(verification)
projects = verification["Project_Name"].nunique()

pending = verification[
    verification["Verification_Result"].str.contains(
        "pending|verification",
        case=False,
        na=False
    )
]

print("RESEARCH EVIDENCE SUMMARY")
print("=" * 45)

print(f"This study documents {total} claims across {projects} lithium projects in Zimbabwe.")

print(
    f"{len(pending)} claims have verification-status labels "
    "indicating outstanding verification work."
)

print(
    "The evidence register distinguishes reported production, "
    "planned capacity, exports, and forecasts."
)

print(
    "These results describe the current evidence audit, "
    "not independently established project performance."
)
```

```python
# Step 4: Create a research findings table

findings = verification[
    [
        "Observation_ID",
        "Project_Name",
        "Claim",
        "Verification_Result"
    ]
].copy()

findings = findings.sort_values(
    by=["Project_Name", "Observation_ID"]
)

print("ZIMBABWE LITHIUM — PRELIMINARY EVIDENCE FINDINGS")
print("=" * 55)

display(findings)

print("\nTotal documented claims:", len(findings))
```

```python
# Step 5: Generate a preliminary research report

report = f"""
ZIMBABWE LITHIUM GOVERNANCE
PRELIMINARY EVIDENCE AUDIT REPORT

1. RESEARCH SCOPE

This evidence audit examines four lithium projects:
Arcadia, Bikita, Sabi Star, and Sandawana.

Total documented claims: {len(verification)}

2. PRELIMINARY OBSERVATIONS

Arcadia:
The register includes claims concerning concentrate
exports and lithium sulphate processing capacity.

Bikita:
The register includes claims concerning ore-processing
capacity, concentrate capacity, and exports.

Sabi Star:
The register currently contains one audited claim
concerning ore-processing capacity.

Sandawana:
The register distinguishes reported FY2025 production
from management forecasts for FY2026.

3. EVIDENCE LIMITATIONS

Several claims require primary-source confirmation,
independent corroboration, or exact passage verification.

Reported capacity, actual production, exports, and
forecasts are different indicators and must not be
treated as interchangeable.

4. RESEARCH RELEVANCE

These observations provide a starting point for examining
industrial development, beneficiation, and the governance
of lithium extraction in Zimbabwe.

They do not yet establish conclusions about local value
capture, foreign dependence, or climate justice.
"""

with open("Zimbabwe_Lithium_Preliminary_Report.txt", "w") as file:
    file.write(report)

print(report)
print("\nReport saved successfully!")
```

```python
# Step 6: Add methodology to the research report

methodology = """
5. RESEARCH METHODOLOGY

This study uses a structured, project-level
evidence-auditing approach.

Four Zimbabwean lithium projects were selected:
Arcadia, Bikita, Sabi Star, and Sandawana.

The research database records:
- Project names and observation IDs
- Production, capacity, and export claims
- Original sources and available supporting sources
- Verification status and research notes

Python and Pandas were used to organize,
inspect, and summarize the evidence.

The analysis distinguishes:
- Reported actual production
- Design or planned capacity
- Export quantities
- Management forecasts

Verification remains provisional.
Claims requiring further source examination
are explicitly identified.
"""

report = report + "\n" + methodology

with open(
    "Zimbabwe_Lithium_Preliminary_Report.txt",
    "w",
    encoding="utf-8"
) as file:
    file.write(report)

print(methodology)
print("\nMethodology added successfully!")
```

```python
# Step 7: Add conclusion and save the updated report

conclusion = """
6. CONCLUSION AND NEXT STEPS

This preliminary audit documents 13 claims
across four lithium projects in Zimbabwe.

The analysis highlights three priorities:

1. Verify production and export figures
   against original company or government records.

2. Distinguish operational output from
   processing capacity and future projections.

3. Expand evidence collection on domestic
   beneficiation, local economic benefits,
   and environmental and community impacts.

The current dataset is an initial research
resource rather than a definitive assessment
of Zimbabwe's lithium sector.

Further verification and analysis are needed
before drawing conclusions about green
extractivism and climate justice.
"""

report = report + "\n" + conclusion

with open(
    "Zimbabwe_Lithium_Preliminary_Report.txt",
    "w",
    encoding="utf-8"
) as file:
    file.write(report)

print(conclusion)
print("\nUpdated report saved successfully!")
```
