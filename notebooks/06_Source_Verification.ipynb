```python
# Step 29: Check missing information and duplicate records

print("DATA QUALITY REPORT")
print("=" * 45)

print("Total records:", len(verification))

print("\nMissing values by column:")
print(verification.isna().sum())

print("\nDuplicate observation IDs:")
print(verification["Observation_ID"].duplicated().sum())

print("\nRecords by project:")
print(verification["Project_Name"].value_counts())

print("\nMissing independent source URLs:")
print(
    verification["Independent_Source_URL"]
    .fillna("")
    .eq("")
    .sum()
)
```

```python
# Step 30: Find incomplete source URLs

url_columns = [
    "Original_Source_URL",
    "Independent_Source_URL"
]

for column in url_columns:
    missing = (
        verification[column]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
    )

    print(f"\n{column}: {missing.sum()} missing")

    display(
        verification.loc[
            missing,
            ["Observation_ID", "Project_Name", "Claim"]
        ]
    )
```
